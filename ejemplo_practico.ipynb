## 11. Manejo de Errores con try/except/finally

````python
try:
    conn = psycopg2.connect(
        host='localhost',
        database='tienda_online',
        user='postgres',
        password='tu_password',
        port='5432'
    )
    
    cursor = conn.cursor()
    
    cursor.execute("SELECT * FROM productos WHERE id = 1")
    resultado = cursor.fetchone()
    print('Producto encontrado:', resultado)
    
    conn.commit()
    print('Cambios guardados con commit()')
    
except psycopg2.Error as e:
    print('Error de base de datos:', e)
    if conn:
        conn.rollback()
        print('Cambios deshechos con rollback()')
    
finally:
    if cursor:
        cursor.close()
    if conn:
        conn.close()
    print('Recursos cerrados')
````

### 📝 Pregunta: ¿Qué pasa si no usamos try/except?
El programa se detiene al primer error y no se liberan los recursos.

### 📝 Pregunta: ¿Por qué usar commit() y rollback()?
- `commit()`: guarda los cambios en la base de datos
- `rollback()`: deshace los cambios si hay error
